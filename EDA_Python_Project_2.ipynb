# Pivot

```python
#pivot_table() creates a summary table showing total Amount 💰 by State 🏙️ and Gender 👨👩 using sum ➕.

New_pivot = df.pivot_table(
    values="Amount",
    columns="Gender",
    index="State",
    aggfunc="min"
)
```

```python
New_pivot
```

```python
#Creates a heatmap to show state-wise amount and gender data with values and colors. 📊🎨

plt.figure(figsize=(10,5))

sns.heatmap(New_pivot,annot=True,fmt=".0f",cmap="YlGnBu")

plt.title("State Wise Amount and Gender",fontsize=18,fontweight="bold",color="black")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
#Creates a pivot table showing the total orders by state and gender group using sum. 📊

New_pivot1 = df.pivot_table(
    values="Orders",
    columns="Gender",
    index="State",
    aggfunc="sum"
)
```

```python
New_pivot1
```

```python
#Creates a heatmap to display state-wise orders and gender data with values shown inside the chart. 📊✨

plt.figure(figsize=(10,5))

sns.heatmap(New_pivot1,annot=True,fmt=".0f")

plt.title("State Wise Orders And Gender",fontsize=18,fontweight="bold",color="black")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
New_pivot2 = df.pivot_table(
    values="Amount",
    columns="Gender",
    index="Orders",
    aggfunc="max"
)
```
